```python
text = "hello transformer"
characters = sorted(set(text))

char_to_id = {
    char : index 
    for index , char in enumerate(characters)
}

# convert each character to its id 

encoded_text = [
    char_to_id[char] 
    for char in text
]
print("Original text:")
print(text)

print("\nEncoded text:")
print(encoded_text)

print("\nNumber of tokens:")
print(len(encoded_text))
```

```python
text = "hello transformer"

characters = sorted(set(text))

char_to_id = {
    char: index
    for index, char in enumerate(characters)
}

id_to_char = {
    index: char
    for index, char in enumerate(characters)
}

# Encoding
encoded = [
    char_to_id[char]
    for char in text
]

# Decoding
decoded = "".join(
    id_to_char[token_id]
    for token_id in encoded
)

print("Original:", text)
print("Encoded:", encoded)
print("Decoded:", decoded)

print("\nRound-trip successful:", decoded == text)
```
